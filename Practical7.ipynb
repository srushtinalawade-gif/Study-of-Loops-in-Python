```python
#Reverse a number
n=int(input("Enter a Number:"))
i=0

while(n>0):
  dig=n%10
  i=i*10+dig
  n=n//10
print("Reversed number:", i)
```

```python
#Check Palindrome  number
n=int(input("Enter a Number:"))
x=n
i=0

while(n>0):
  dig=n%10
  i=i*10+dig
  n=n//10
if (x==i):
  print("The number is a palindrome.")
else:
  print("The number is not a palindrome.")
```

```python
s = str(input("Enter a word:"))
i, j=0, len(s) - 1
is_palindrome = True

while i<j:
  if s[i] !=s[j]:
    is_palindrome= False
    break
  i +=1
  j -=1

if is_palindrome:
  print("yes")
else:
  print("no")

```

```python
# Multiplication of Two 3x3 Matrices (Most Important- Viva)


A=[[1,2,3],[1,2,1],[7,4,1]]
B=[[0,1,2],[2,3,8],[4,5,7]]
Result=[[0,0,0],[0,0,0],[0,0,0]]

num_col=len(A[0])
num_rows=len(B)

if num_col==num_rows:
  for i in range(3):
    for j in range(3):
      for k in range(3):
        Result[i][j] += A[i][k] * B[k][j]
  print('Product:')
  for row in Result:
    print(row)
else:
  print('Matrices cant be multiplied')
```

```python
#Python Program to Accept Three Digit Numbers and Print all possible combinations

d1= int(input("Enter first digit:"))
d2= int(input("Enter second digit:"))
d3= int(input("Enter third digit:"))

d=[d1,d2,d3]

for i in range(0,3):
  for j in range(0,3):
    for k in range(0,3):
      if (d[i]!=d[j] and d[j]!=d[k] and d[k]!=d[i]):
        print(d[i],d[j],d[k])
```

```python
#Python Program Armstrong
N=input('Enter a number: ')
total=0

for i in N:
  total+= int(i)**len(N)

if total==int(N):
  print(N," is an Armstrong Number")
else:
  print(N," is not an Armstrong Number")
```
